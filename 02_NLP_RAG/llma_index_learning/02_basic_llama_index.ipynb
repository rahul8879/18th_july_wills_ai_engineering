```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from llama_index.core import Settings
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.llms.openai import OpenAI
Settings.embed_model = HuggingFaceEmbedding(model_name="sentence-transformers/all-MiniLM-L6-v2",)
Settings.llm = OpenAI(model="gpt-4o-mini")
```

```python
from llama_index.core import (
    SimpleDirectoryReader,
    VectorStoreIndex,
    SummaryIndex,
    KeywordTableIndex,
    TreeIndex
)

from llama_index.core.response.pprint_utils import pprint_response
```

```python
documents = SimpleDirectoryReader('data_test').load_data()
index = VectorStoreIndex(documents) # chunking ---> embedding
query_engine = index.as_query_engine() 
response = query_engine.query("What is a personal loan?")# full RAG ===questions --> vector --> search in index --top 3/5-->context -->llm --final reponse

```

```python
pprint_response(response,show_source=True)
```

```python
from llama_index.core.node_parser import SentenceSplitter

splitter = SentenceSplitter(
    chunk_size=1000,
    chunk_overlap=50
)

```

```python
documents = SimpleDirectoryReader('data_test').load_data()
nodes = splitter.get_nodes_from_documents(documents)

#
vector_index = VectorStoreIndex(nodes)
summary_index = SummaryIndex(nodes)
keyword_table_index = KeywordTableIndex(nodes)
tree_index = TreeIndex(nodes)

rv =vector_index.as_query_engine()
sv = summary_index.as_query_engine()
kv = keyword_table_index.as_query_engine()
tv = tree_index.as_query_engine()
```

```python
q = "How many days is maternity leave?"
print("vector index")
pprint_response(rv.query(q),show_source=True)
```

```python
print("keyword index")
pprint_response(kv.query(q),show_source=True)
```

```python
print("tree index")
pprint_response(tv.query(q),show_source=True)
```
