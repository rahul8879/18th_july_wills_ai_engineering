```python

# langchain document loader
# txt
# csv
# pdf
# ppt
# docs
# any documents

from langchain_community.document_loaders import PyPDFLoader
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_text_splitters import RecursiveCharacterTextSplitter

from dotenv import load_dotenv
load_dotenv()
```

```python
loader = PyPDFLoader('bajaj_finance_policy_prose_v1.pdf')
```

```python
pages = loader.load()
```

```python
len(pages)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=350,
                                          separators=["\n\n", "\n", " ", ""],
                                           chunk_overlap=0)


chunks = splitter.split_documents(pages)
```

```python
len(chunks)
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
chunk_text = [chunk.page_content for chunk in chunks]
chunk_meta = [chunk.metadata for chunk in chunks]
```

```python
chunk_vectors = model.encode(
    chunk_text,
    #  show_progress_bar=True,
    # convert_to_numpy=True,
    # normalize_embeddings=True   # we will discuss on this w=in fine tuning chapter
)
    
```

```python
chunk_vectors
```

```python
store = {
    'vectors':chunk_vectors,
    'text':chunk_text,
    'meta':chunk_meta,
    'model':'all-MiniLM-L6-v2'
}
```

```python

import pickle

with open("bfl_store.pkl", "wb") as f:
    pickle.dump(store, f)
```
