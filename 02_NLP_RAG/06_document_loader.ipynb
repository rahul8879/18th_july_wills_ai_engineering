```python
# chunk ==> size ?? 

# fixed len

dummy_text = "This is a sample text that we will use to demonstrate text chunking. We will split this text into smaller chunks based on a specified chunk size. Each chunk will contain a fixed number of characters, and we will ensure that the chunks do not exceed the specified size. This approach is useful for processing large texts in manageable pieces."
```

```python
def chunk_v0(text, chunk_size=50):
    chunks = []
    for i in range(0, len(text), chunk_size):
        chunk = text[i:i + chunk_size]
        chunks.append(chunk)
    return chunks
chunk_v0(dummy_text)


# langchain ===> chunking ===> fixed len == text_splitter
```

```python
len(chunk_v0(dummy_text)[0])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Prepare your raw text document
long_document = """
LangChain is a framework designed to simplify the creation of applications using LLMs.
Text splitting is a critical component of building an effective retrieval pipeline.

By breaking text down recursively, we ensure that paragraphs and sentences remain intact
wherever possible, allowing the model to grasp the context fully.
"""

# 2. Initialize the splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    length_function=len
)

# 3. Create chunks (returns a list of strings)
chunks = text_splitter.split_text(long_document)

# Preview chunks
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:\n{chunk}\n{'-'*20}")
```
